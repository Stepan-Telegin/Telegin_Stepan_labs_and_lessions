**Задание 2.** Создайте двумерный массив размерности `n*m` (`n` и `m` вводятся с клавиатуры) и заполните его случайными значениями. Найдите в этом массиве:

1. Все элементы, большие некоторого числа `k` (вводится с клавиатуры).
2. Статистические характеристики элементов массива (математическое ожидание, дисперсию, среднее квадратичное отклонение).
3. Минимальный и максимальный элементы массива.




```python
n = int(input("Введите n: "))
m = int(input("Введите m: "))
k = float(input("Введите k: "))

random_matrix = np.random.random((n, m)) * 100
print(random_matrix)
```

```python
# 1 пункт
greater_than = random_matrix[random_matrix > k]
print(greater_than)
```

```python
# 2 пункт
mean_value = random_matrix.mean()
variance_value = random_matrix.var()
std_value = random_matrix.std()

print(mean_value)
print(variance_value)
print(std_value)
```

```python
# 3 пункт
minimum = random_matrix.min()
maximum = random_matrix.max()

print(minimum)
print(maximum)
```

**Задание 3** Даны два вектора одинакового размера. Найдите:

*  Косинус угла между векторами.
*  Евклидово расстояние между векторами.

```python
# решение задания 3
vector_a = np.array([1, 2, 3, 4])
vector_b = np.array([5, 6, 7, 8])

cosine = np.dot(vector_a, vector_b) / (np.linalg.norm(vector_a) * np.linalg.norm(vector_b))
distance = np.sqrt(((vector_a - vector_b) ** 2).sum())

print(cosine)
print(distance)
```
